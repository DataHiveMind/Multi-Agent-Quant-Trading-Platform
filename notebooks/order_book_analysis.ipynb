```python
# notebooks/order_book_analysis.ipynb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Simulated bid-ask spread analysis
order_book = pd.read_csv("datasets/order_book_data.csv")

# Compute bid-ask spread
spread = order_book["Ask"] - order_book["Bid"]

plt.figure(figsize=(12, 6))
plt.plot(order_book["Time"], spread, label="Bid-Ask Spread")
plt.xlabel("Time")
plt.ylabel("Spread")
plt.title("Order Book Dynamics Analysis")
plt.legend()
plt.show()
```
