```python
# notebooks/exploratory_analysis.ipynb
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("datasets/market_data.csv")

# Plot historical price series
plt.figure(figsize=(12, 6))
plt.plot(df["Date"], df["Close"], label="Closing Prices")
plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Market Data Exploration")
plt.legend()
plt.show()
```
