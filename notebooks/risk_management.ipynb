```python
# notebooks/risk_management.ipynb
import numpy as np
import matplotlib.pyplot as plt

# Simulate portfolio return distribution
num_simulations = 10000
portfolio_returns = np.random.normal(loc=0.05, scale=0.1, size=num_simulations)

plt.figure(figsize=(10, 6))
plt.hist(portfolio_returns, bins=50, alpha=0.7, color="blue")
plt.title("Monte Carlo Simulated Portfolio Returns")
plt.xlabel("Returns")
plt.ylabel("Frequency")
plt.show()
```
