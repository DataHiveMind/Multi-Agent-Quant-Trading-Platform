```python
# notebooks/quantitative_research.ipynb
import pandas as pd
import numpy as np
import statsmodels.api as sm

# Load factor data
df = pd.read_csv("datasets/factor_data.csv")

# Perform linear regression for factor analysis
X = df[["Momentum", "Value", "Quality"]]
y = df["Returns"]

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()

# Show model summary
print(model.summary())
```
